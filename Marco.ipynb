```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("AIDC-AI/Marco-o1")
model = AutoModelForCausalLM.from_pretrained("AIDC-AI/Marco-o1")
```

```python
messages = [
    {"role": "user", "content": "தமிழ் இலக்கியம் இரண்டாயிரம் ஆண்டுகளுக்கும் மேலான பழமையான வரலாறு கொண்டது"},
]
inputs = tokenizer.apply_chat_template(
	messages,
	add_generation_prompt=True,
	tokenize=True,
	return_dict=True,
	return_tensors="pt",
).to(model.device)

outputs = model.generate(**inputs, max_new_tokens=1000)
print(tokenizer.decode(outputs[0][inputs["input_ids"].shape[-1]:]))
```

```python
# Few-shot prompt
prompt = """I will give analogies in the format A : B :: C : D.
A is an English word, B is its Tamil translation, and you must give the Tamil word for D.

Examples:
Father : அப்பா :: Mother : அம்மா
Sun : சூரியன் :: Moon : சந்திரன்
King : ராஜா :: Queen : ராணி
Now complete this analogy:
Animal : குதிரை :: Bird :கோழி 
மாடு : Animal  :: புறா:"""

# Tokenize
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Generate full analogy
outputs = model.generate(
    **inputs,
    max_new_tokens=1000,    # Enough to complete the analogy
    do_sample=False
)

# Decode
prediction = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(prediction)
```
