```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("ai4bharat/Airavata")
model = AutoModelForCausalLM.from_pretrained("ai4bharat/Airavata")
```

```python
# Prepare input
prompt = "King is to Queen as Raja is to ?"  # English-Tamil analogy
inputs = tokenizer(prompt, return_tensors="pt")

# Generate output
with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=20,
        temperature=0.7,
        do_sample=True,
        top_k=50
    )

# Decode
response = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(response)
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForMaskedLM

tokenizer = AutoTokenizer.from_pretrained("ai4bharat/IndicBERTv2-SS")
model = AutoModelForMaskedLM.from_pretrained("ai4bharat/IndicBERTv2-SS")
```

```python
analogy_list = [
    "சூரியன் : பகல் :: சந்திரன் :  [MASK]"
]

for sentence in analogy_list:
    inputs = tokenizer(sentence, return_tensors="pt")
    outputs = model(**inputs)
    predictions = outputs.logits
    mask_token_index = (inputs.input_ids == tokenizer.mask_token_id)[0].nonzero(as_tuple=True)[0]
    top_tokens = torch.topk(predictions[0, mask_token_index], 5, dim=1).indices[0].tolist()
    print(f"Input: {sentence}")
    print("Prediction:", [tokenizer.decode([token]) for token in top_tokens])
    print()
```

```python
# Load model directly
from transformers import AutoModel ,AutoTokenizer
model = AutoModel.from_pretrained("ai4bharat/indic-bert", torch_dtype="auto").to("cuda:0")
tokenizer = AutoTokenizer.from_pretrained("ai4bharat/indic-bert", torch_dtype="auto").to("cuda:0")
```

```python
# Get embeddings of each word separately
def get_embedding(word):
    inputs = tokenizer(word, return_tensors="pt")
    with torch.no_grad():
        outputs = model(**inputs)
        return outputs.last_hidden_state.mean(dim=1).squeeze()

# Tamil analogy: அப்பா : ஆண் :: அம்மா : ?
appa = get_embedding("அப்பா")
aan = get_embedding("ஆண்")
amma = get_embedding("அம்மா")

# Do the analogy vector calculation
result_vector = amma + (aan - appa)

# Find closest word (cosine similarity search)
from torch.nn.functional import cosine_similarity

vocab = list(tokenizer.vocab.keys())
similarities = []
for word in vocab:
    emb = get_embedding(word)
    sim = cosine_similarity(result_vector, emb.unsqueeze(0)).item()
    similarities.append((word, sim))

# Top 10 most similar tokens
top_words = sorted(similarities, key=lambda x: x[1], reverse=True)[:10]
for word, score in top_words:
    print(f"{word}: {score:.4f}")
```
