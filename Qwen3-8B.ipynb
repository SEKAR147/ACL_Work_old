# Loading Qwen Model

```python
import analogy_dataset as ad
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
from tqdm import tqdm
import pandas as pd

#  Step 1: Load Qwen properly
model_id = "Qwen/Qwen3-8B"

tokenizer = AutoTokenizer.from_pretrained(model_id)

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.float16
).to("cuda") 

print("Model loaded on:", next(model.parameters()).device)
print("cuda available" if torch.cuda.is_available() else "CPU")
```

# Starting Full Translation

```python


#  Step 2: Prepare prompt builder
def build_prompt(row):
    return f"""You are a multilingual translator. Translate this all analogy MCQ from Hindi to Tamil.

Question: {row['Part1']} : {row['Part2']} :: {row['Part3']} : ?
Options:
A. {row['Option_1']}
B. {row['Option_2']}
C. {row['Option_3']}
D. {row['Option_4']}
"""

# Step 3: Translation using the loaded Qwen (chat-style model)
def get_translation(prompt):
    messages = [{"role": "user", "content": prompt}]
    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        return_tensors="pt",
        return_dict=True,
        tokenize=True,
        enable_thinking = False
    )
    inputs = {k: v.to(model.device) for k, v in inputs.items()}
    outputs = model.generate(**inputs, max_new_tokens=512)
    return tokenizer.decode(outputs[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True).strip()

all_rows = []
iterator = ad.get_iterator()

while True:
    row = iterator.get_next_row()
    if not row:
        break
    all_rows.append(row)

batch_size = 100
total = len(all_rows)

# Run this for any batch index you want
batch_index = 1  # or 1, 2, 3...

start = batch_index * batch_size
end = min(start + batch_size, total)

batch_rows = all_rows[start:end]

translated_data = []

for row in batch_rows:
    prompt = build_prompt(row)
    try:
        tamil = get_translation(prompt)
    except Exception as e:
        tamil = f"ERROR: {e}"
        print(f"Error at row: {row['Formatted_Question']} — {e}")

    translated_data.append({
        "Formatted_Question": row["Formatted_Question"],
        "Answer_Hindi": row["All_Correct_Answers"],
        "Tamil_Translated": tamil
    })

# Save per batch
df = pd.DataFrame(translated_data)
df.to_csv(f"tats_batch_{batch_index}.csv", index=False)
print(f" Batch {batch_index} saved!")



# #  Step 4: Run through analogy dataset
# iterator = ad.get_iterator()
# translated_data = []

# while True:
#     row = iterator.get_next_row()
#     if not row:
#         break

#     prompt = build_prompt(row)
#     try:
#         tamil = get_translation(prompt)
#     except Exception as e:
#         tamil = f"ERROR: {e}"
#         print(f" Error at row: {row['Formatted_Question']} — {e}")

#     translated_data.append({
#         "Formatted_Question": row["Formatted_Question"],
#         "Answer_Hindi": row["All_Correct_Answers"],
#         "Tamil_Translated": tamil
#     })
  
# # Step 5: Save to file
# df = pd.DataFrame(translated_data)
# df.to_csv("tats_translated_from_package.csv", index=False)
# print(" All translated outputs saved to: tats_translated_from_package.csv")
```

```python
# import pandas as pd
# from transformers import AutoTokenizer, AutoModelForCausalLM
# import torch
# from tqdm import tqdm

# #  Load model
# model_id = "Qwen/Qwen3-8B"
# tokenizer = AutoTokenizer.from_pretrained(model_id, trust_remote_code=True)
# model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.float16, device_map="auto")

#  Reverse translation function
def translate_to_hindi(tamil_text):
    prompt = f"""You are a professional multilingual translator. Translate the following Tamil analogy question back to Hindi.
Tamil Input:
{tamil_text}
Respond only with the translated Hindi version. Do not explain."""
    
    messages = [{"role": "user", "content": prompt}]
    
    try:
        # apply_chat_template returns just input_ids (a tensor), not a dict
        inputs = tokenizer.apply_chat_template(
            messages,
            add_generation_prompt=True,
            return_tensors="pt",
            tokenize=True,
            enable_thinking=False
        )
        
        # Move tensor to model's device
        inputs = inputs.to(model.device)
        
        with torch.no_grad():
            output_ids = model.generate(
                input_ids=inputs,  # Just pass the tensor directly
                max_new_tokens=256,
                pad_token_id=tokenizer.pad_token_id
            )
        
        # Decode the output
        full_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
        
        # Remove the prompt part to get only the response
        prompt_text = tokenizer.decode(inputs[0], skip_special_tokens=True)
        return full_text.replace(prompt_text, "").strip()
        
    except Exception as e:
        return f"ERROR: {str(e)}"
#  Load your file
df = pd.read_csv("tats_batch_1.csv")

#  Translate each row
reversed_rows = []
for _, row in tqdm(df.iterrows(), total=len(df)):
    tamil = row["Tamil_Translated"]
    hindi = translate_to_hindi(tamil)
    reversed_rows.append({
        "Tamil_Translated": tamil,
        "Original_Hindi": row["Formatted_Question"],
        "Reversed_Hindi": hindi,
        "Answer_Hindi": row["Answer_Hindi"]
    })

#  Save output
out_df = pd.DataFrame(reversed_rows)
out_df.to_csv("tats_batch_1_reverse.csv", index=False)
print(" Reverse-translated file saved as tats_batch_1_reverse.csv")
```

```python
import pandas as pd

# Load your reverse word-by-word output CSV
df = pd.read_csv("tats_batch_1_reverse.csv")  # Update filename if needed

# Function to clean text (normalize punctuation, case, etc.)
def clean(text):
    return str(text).strip().lower().replace("।", "").replace(".", "").replace("?", "")

# Compute exact match between original and reversed
df["Exact_Match"] = df.apply(
    lambda row: clean(row["Reversed_Hindi"]) == clean(row["Original_Hindi"]),
    axis=1
)

# Save evaluation CSV
df.to_csv("reverse_wordwise_batch_1_eval.csv", index=False)

# Summary of accuracy
correct = df["Exact_Match"].sum()
total = len(df)
print(f"Exact Match Accuracy: {correct}/{total} = {correct/total:.2%}")


```

```python
#Claude Code
# import pandas as pd
# from transformers import AutoTokenizer, AutoModelForCausalLM
# import torch
# from tqdm import tqdm
# import analogy_dataset as ad

# # Load model
# model_id = "Qwen/Qwen3-8B"
# tokenizer = AutoTokenizer.from_pretrained(model_id, trust_remote_code=True)
# model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.float16, device_map="auto")
# print(f"Model loaded on: {model.device}")

# Improved prompt builder for accurate translation
def build_prompt(row):
    return f"""You are a professional Hindi to Tamil translator. Translate the following analogy question accurately.

INSTRUCTIONS:
1. Translate each Hindi word to its Tamil equivalent
2. Maintain the same structure and word order
3. Keep punctuation marks (::, ?, -) as they are
4. Translate all 4 options (A, B, C, D)
5. Output ONLY the Tamil translation - no explanations, no original Hindi text

Hindi Question:
{row['Part1']} : {row['Part2']} :: {row['Part3']} : ?
विकल्प:
A. {row['Option_1']}
B. {row['Option_2']}
C. {row['Option_3']}
D. {row['Option_4']}

Tamil Translation:"""

# Translation function with better parameters
def get_translation(prompt):
    messages = [{"role": "user", "content": prompt}]
    
    try:
        inputs = tokenizer.apply_chat_template(
            messages,
            add_generation_prompt=True,
            return_tensors="pt",
            tokenize=True,
            enable_thinking=False
        )
        
        # Move to device
        inputs = inputs.to(model.device)
        
        with torch.no_grad():
            outputs = model.generate(
                input_ids=inputs,
                max_new_tokens=800,  # Increased for full question + options
                pad_token_id=tokenizer.pad_token_id,
                temperature=0.3,  # Low temperature for consistent translation
                do_sample=True,  # Some sampling for natural translation
                top_p=0.9,
                repetition_penalty=1.1
            )
        
        # Decode only the generated part
        full_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
        prompt_text = tokenizer.decode(inputs[0], skip_special_tokens=True)
        tamil_output = full_text.replace(prompt_text, "").strip()
        
        # Clean any artifacts
        tamil_output = tamil_output.replace("Tamil Translation:", "").strip()
        tamil_output = tamil_output.replace("Translation:", "").strip()
        
        return tamil_output
        
    except Exception as e:
        return f"ERROR: {str(e)}"

# Load all rows from dataset
print("Loading dataset...")
all_rows = []
iterator = ad.get_iterator()
while True:
    row = iterator.get_next_row()
    if not row:
        break
    all_rows.append(row)

print(f"Total rows in dataset: {len(all_rows)}")

# Batch configuration
batch_size = 100
batch_index = 0  # Change this for different batches: 0, 1, 2, 3, 4...

start = batch_index * batch_size
end = min(start + batch_size, len(all_rows))
batch_rows = all_rows[start:end]

print(f"\nProcessing Batch {batch_index}: Rows {start} to {end-1}")
print("="*100)

# Translate batch
translated_data = []

for idx, row in enumerate(tqdm(batch_rows, desc=f"Batch {batch_index}")):
    prompt = build_prompt(row)
    
    try:
        tamil = get_translation(prompt)
    except Exception as e:
        tamil = f"ERROR: {e}"
        print(f"\n✗ Error at row {start + idx}: {e}")
    
    translated_data.append({
        "Row_Index": start + idx,
        "Formatted_Question": row["Formatted_Question"],
        "Answer_Hindi": row["All_Correct_Answers"],
        "Tamil_Translated": tamil
    })
    
    # Show sample every 20 rows
    if idx % 20 == 0 and idx > 0:
        print(f"\n[Sample Translation - Row {start + idx}]")
        print(f"Hindi:  {row['Formatted_Question'][:80]}...")
        print(f"Tamil:  {tamil[:80]}...")
        print("-"*100)

# Save batch
output_file = f"tats_batch_{batch_index}.csv"
df = pd.DataFrame(translated_data)
df.to_csv(output_file, index=False)

print(f"\n{'='*100}")
print(f"✓ Batch {batch_index} completed!")
print(f"✓ Saved to: {output_file}")
print(f"✓ Rows processed: {len(translated_data)}")
print(f"{'='*100}")

# Quick quality check
error_count = df['Tamil_Translated'].str.contains('ERROR', na=False).sum()
if error_count > 0:
    print(f"⚠ Warning: {error_count} rows had errors")
else:
    print(f"✓ All translations completed successfully!")
```
