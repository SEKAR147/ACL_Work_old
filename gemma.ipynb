```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("google/gemma-7b")
model = AutoModelForCausalLM.from_pretrained("google/gemma-7b", device_map="auto", torch_dtype=torch.float16)


```

```python
# Few-shot prompt
prompt = """I will give analogies in the format A : B :: C : D.
A is an English word, B is its Tamil translation, and you must give the Tamil word for D.

Examples:
Father : அப்பா :: Mother : அம்மா
Sun : சூரியன் :: Moon : சந்திரன்
King : ராஜா :: Queen : ராணி
Now complete this analogy:
Animal : குதிரை :: Bird :கோழி 
மாடு : Animal  :: புறா:?"""

# Tokenize
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Generate full analogy
outputs = model.generate(
    **inputs,
    max_new_tokens=10,    # Enough to complete the analogy
    do_sample=False
)

# Decode
prediction = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(prediction)
```

```python
import pandas as pd


# Example data
data = [
    ("Father", "அப்பா", "Mother", "அம்மா"),
    ("Sun", "சூரியன்", "Moon", "சந்திரன்"),
    ("Dog", "நாய்", "Cat", "பூனை"),
    ("Water", "தண்ணீர்", "Fire", "தீ")
]

few_shot = """I will give analogies in the format A : B :: C : D.
A is an English word, B is its Tamil translation, and you must give the Tamil word for D.

Examples:
Father : அப்பா ::  Mother : அம்மா
Sun : சூரியன் ::   Moon :    சந்திரன்
King : ராஜா :: Queen : ராணி
Father : அப்பா :: Mother : அம்மா
Sun : சூரியன் :: Moon : சந்திரன்
King : ராஜா :: Queen : ராணி
Dog : நாய் :: Cat : பூனை
Tree : மரம் :: Flower : மலர்
Fire : தீ :: Water : 

"""

results = []

for engA, tamB, engC, goldD in data:
    prompt = few_shot + f"\nNow complete:\n{engA} : {tamB} :: {engC} :"
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(**inputs, max_new_tokens=4, do_sample=False)
    prediction = tokenizer.decode(outputs[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True).strip()
    
    is_correct = prediction == goldD
    results.append((engA, tamB, engC, goldD, prediction, is_correct))

# Convert to DataFrame
df = pd.DataFrame(results, columns=["Eng_A", "Tam_B", "Eng_C", "Gold_D", "Predicted", "Correct"])

print(df)
print(f"Accuracy: {df['Correct'].mean()*100:.2f}%")
```
